## 1. 데이터프레임 기본

데이터프레임은 행과 열로 구성된 2차원 표 형태의 자료구조이다.

엑셀 시트처럼 생겼고, 각 열마다 다른 자료형을 가질 수 있다.

주요 확인 메서드
- df.shape : 행수, 열수 확인
- df.dtypes : 각 열의 자료형 확인
- df.head() : 상위 5행 확인
- df.describe() : 수치형 열 기초 통계

```python
import pandas as pd

df = pd.DataFrame({
    'name'  : ['Kim', 'Lee', 'Park', 'Choi', 'Jung'],
    'age'   : [25, 30, 22, 28, 35],
    'city'  : ['Seoul', 'Busan', 'Seoul', 'Incheon', 'Busan'],
    'score' : [88, 92, 75, 96, 81]
})

df
```

```python
print(df.shape)
print()
print(df.dtypes)
```

```python
df.describe()
```

## 2. loc 조건문

loc는 라벨(이름) 기반으로 행과 열을 선택하는 방법이다.

사용 형태: df.loc[행 조건, 열 이름]

- 행 조건 자리에 True/False 배열을 넣으면 조건 필터링이 된다.
- 열 이름을 생략하면 모든 열이 반환된다.
- iloc은 위치(숫자) 기반, loc은 라벨(이름) 기반이라는 점이 다르다.

```python
# score가 85점 이상인 행 전체 선택
df.loc[df['score'] >= 85]
```

```python
# score가 85점 이상인 행에서 name, score 열만 선택
df.loc[df['score'] >= 85, ['name', 'score']]
```

```python
# city가 Seoul이고 score가 80 이상인 행
df.loc[(df['city'] == 'Seoul') & (df['score'] >= 80)]
```

```python
# loc으로 특정 인덱스의 특정 열 값 수정
df.loc[0, 'score'] = 90
df
```

## 3. plt.plot() 과 plt.show()

matplotlib은 파이썬 기본 시각화 라이브러리이다.

plt.plot(x, y) : 선 그래프를 그린다.

plt.show() : 그래프를 화면에 출력한다. 주피터에서는 없어도 보이지만 명시적으로 쓰는 것이 좋다.

자주 쓰는 옵션
- color : 선 색상
- marker : 데이터 포인트 모양 (o, s, ^ 등)
- linestyle : 선 스타일 (solid, dashed, dotted)
- label : 범례 이름
- plt.title() : 제목
- plt.xlabel(), plt.ylabel() : 축 이름
- plt.legend() : 범례 표시

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 30, 22]

plt.plot(x, y)
plt.title('Basic Line Chart')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# marker와 color 옵션 추가
plt.plot(x, y, color='steelblue', marker='o', linestyle='dashed', label='value')
plt.title('Line Chart with Options')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
# 데이터프레임 데이터로 시각화
plt.plot(df['name'], df['score'], marker='o', color='tomato', label='score')
plt.title('Score by Name')
plt.xlabel('Name')
plt.ylabel('Score')
plt.legend()
plt.show()
```

## 4. 종합 예제

```python
# loc으로 필터링한 결과를 바로 시각화

high = df.loc[df['score'] >= 85, ['name', 'score']]

plt.plot(high['name'], high['score'], marker='s', color='seagreen', label='85점 이상')
plt.title('High Score Students')
plt.xlabel('Name')
plt.ylabel('Score')
plt.legend()
plt.show()
```
